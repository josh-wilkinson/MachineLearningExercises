```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
import os
```

# Examine the data

```python
# loda the dataset
data_path = os.path.join(os.getcwd(),"listings.csv")
df = pd.read_csv(data_path)
```

```python
# get the info
print(df.info())
```

As one can see the price in only present for 6321 out of 10000 rows.

```python
df["price"].describe()
```

For this reason I will first remove the rows without price and then split the data into training and test set.

```python
df['price'].notna()
```

```python
# remove rows without price
df = df.dropna(subset=['price'])
print(df.info())
```

```python
# split the dataset into train and test
train_df, test_df = train_test_split(df, test_size=0.2, random_state=1)
# save the train and test datasets
train_df.to_csv(os.path.join(os.getcwd(), "train_listings.csv"), index=False)
test_df.to_csv(os.path.join(os.getcwd(), "test_listings.csv"), index=False)
```
